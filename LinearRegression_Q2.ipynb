```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
re=pd.read_csv("/content/real estate - Sheet1.csv")
re
```

```python
y=re['price']
y
```

```python
x=re.drop("price", axis=1)
x
```

```python
xtrain, xtest, ytrain, ytest=train_test_split(x, y, train_size=0.8, random_state=0)
```

```python
xtrain.shape, xtest.shape, ytrain.shape, ytest.shape
```

```python
LR=LinearRegression()
LR.fit(xtrain, ytrain)
```

```python
ypred=LR.predict(pd.DataFrame(xtest))
```

```python
c = [i for i in range(1, 201)]
fig = plt.figure()
plt.plot(c, ytest, color = "Blue", linewidth = 2, linestyle = "--", label = "Actual")
plt.plot(c, ypred, color = "Black", linewidth = 3, linestyle = "-", label = "Predicted")
plt.legend()
plt.title("Price based on house stats")
plt.xlabel("FACTORS")
plt.ylabel("Price")
plt.show()
```
