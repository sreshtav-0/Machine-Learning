```python
#Merge function:
#It connects columns or indexes in the DataFrame based on one or more keys
Df1=pd.DataFrame({"ID":[1, 2, 3, 4], "Class":[5, 6, 7, 8]})
Df1
```

```python
Df2=pd.DataFrame({"ID":[1, 2, 3, 4], "Name":["A", "B", "C", "D"]})
Df2
```

```python
pd.merge(Df2, Df1, on="ID")
```

```python
pd.merge(Df1, Df1, on="ID")
```

```python
pd.merge(Df1, Df1, on="ID", suffixes=("one", "two"))
```

```python
Df3=pd.DataFrame({"ID":[1, 2, 3], "Name":["A", "B", "C"]})
Df3
```

```python
pd.merge(Df3, Df1, on="ID", how="left")
```

```python
pd.merge(Df3, Df1, on="ID", how="right")
```
