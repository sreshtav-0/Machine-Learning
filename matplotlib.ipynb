```python
z=3*x
a=4*x
```

```python
plt.plot(x, z, "r.--", markersize=15, markerfacecolor="crimson", markeredgecolor="blue", linewidth=2, label="first tornado path")
plt.plot(y, a, "yo-.", markersize=15, markerfacecolor="purple", markeredgecolor="orange", linewidth=3, label="second tornado path")
plt.xlabel("time in half hours")
plt.ylabel("distance")
plt.legend(loc=5)
plt.xlim(2, 5)
plt.ylim(4, 10)
plt.show()
```

```python
fig=plt.figure()
axes=fig.add_axes([0, 0, 1, 1])
#[0, 0, 1, 1] represents the left, bottom, width, height
axes.plot(x, y)
```

```python
fig=plt.figure()
axes1=fig.add_axes([0, 0, 0.5, 0.5])
axes2=fig.add_axes([0.5, 0.5, 0.25, 0.25])
#[0, 0, 1, 1] represents the left, bottom, width, height
axes1.plot(x, y)
axes2.plot(x, z)
axes1.set_xlabel("big figure xlabel")
axes1.set_ylabel("big figure ylabel")
axes1.set_title("Big Figure")
axes2.set_xlabel("small figure xlabel")
axes2.set_ylabel("small figure ylabel")
axes2.set_title("Small Figure")
axes1.set_xlim(1, 5)
axes1.set_ylim(1, 10)
plt.show()
```

```python
fig, axes=plt.subplots(nrows=2, ncols=2)
axes[0][0].plot(x, y)
axes[0][0].set_title("Subplots (0, 0)")
axes[0][0].set_xlabel("0, 0 XLabel")
axes[0][0].set_ylabel("0, 0 YLabel")

axes[0][1].plot(x, y)
axes[0][1].set_title("Subplots (0, 1)")
axes[0][1].set_xlabel("0, 1 XLabel")
axes[0][1].set_ylabel("0, 1 YLabel")

axes[1][0].plot(x, y)
axes[1][0].set_title("Subplots (1, 0)")
axes[1][0].set_xlabel("1, 0 XLabel")
axes[1][0].set_ylabel("1, 0 YLabel")

axes[1][1].plot(x, y)
axes[1][1].set_title("Subplots (1, 1)")
axes[1][1].set_xlabel("1, 1 XLabel")
axes[1][1].set_ylabel("1, 1 YLabel")

fig.subplots_adjust(wspace=0.5, hspace=0.5)
```

#Histograms

```python
ML_students=np.random.randint(14, 45, 100)
ML_students
```

```python
PY_students=np.random.randint(10, 45, 100)
PY_students
```

```python
plt.hist(ML_students)
plt.title("Machine Learning Graph")
plt.xlabel("Age")
plt.ylabel("Level of Proficiency")
```

```python
bins=[15, 20, 25, 30, 35, 40, 45]
plt.hist(ML_students, bins=bins, rwidth=0.5, label="ML students", color="green")
plt.title("Machine Learning Graph")
plt.xlabel("Age")
plt.ylabel("Level of Proficiency")
plt.legend()
```

```python
bins=[15, 20, 25, 30, 35, 40, 45]
plt.hist(PY_students, bins=bins, rwidth=0.5, label="PY students", color="blue")
plt.title("All Student Graph")
plt.xlabel("Age")
plt.ylabel("Level of Proficiency")
plt.legend()
```

#Pie Plot (PieChart)

```python
subjects=["math", "ela", "science", "humanities", "AIML"]
classs=[40, 40, 35, 25, 20]
```

```python
explode=[0, 0, 0.2, 0.2, 0.2]
colors=["red", "blue", "green", "yellow", "gray"]
textprops={"fontsize":15}
plt.pie(classs, labels=subjects, explode=explode, colors=colors, rotatelabels=True, textprops=textprops, radius=2, startangle=90, autopct="%0.2f%%")
plt.show()
```
